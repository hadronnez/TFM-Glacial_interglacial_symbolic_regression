## Julia testing 

#### Julia REPL modes:

##### Julia mode: 
    Writing "julia" in the Powershell. 
    Here, code can be run, similar as writing the command "python" for python.

##### PKG mode:  
    Writing "pkg" in the Powershell or typing "]" in the julia mode. 
    Here, packages and environments can be managed, similar as writing the command "pip" for python.

##### Shell mode:
    Writing "shell" in the Powershell or typing ";" in the julia mode. 
    Here, shell commands can be run but without leaving the julia session.

##### Help mode:
    Writing "help?" in the Powershell or typing "?" in the julia mode.  
    Here, command can be written and documentation of those is provided.
