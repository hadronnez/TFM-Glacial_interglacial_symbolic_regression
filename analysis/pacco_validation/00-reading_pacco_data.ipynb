```julia
using NCDatasets, CairoMakie

ROOT = normpath(joinpath(@__DIR__, "..", "..")) 
data_path = joinpath(ROOT, "pacco_model", "data", "BASE.nc")
output_path = joinpath(ROOT, "analysis", "pacco_validation", "output")

# Load ice volume
time, volume = NCDataset(data_path)["time"][:], NCDataset(data_path)["Vol"][:]

# Plot ice volume
fig = Figure()
ax = Axis(fig[1, 1])
lines!(ax, time, volume)
save(joinpath(output_path, "ice_volume.png"), fig)

# Select time slice
function find_times(t0, tend, time_vector) 
     # findmin(vector)[2] extracts only the index
     t0_index = findmin(abs.(t0 .- time_vector))[2]
     tend_index = findmin(abs.(tend .- time_vector))[2]
     return t0_index, tend_index
end

# -- Select only Late Pleistocene
tspan = find_times(-800e3, 0.0, time)
new_time, new_volume = time[tspan[1]:tspan[2]], volume[tspan[1]:tspan[2]]

# Plot
fig = Figure()
ax = Axis(fig[1, 1])
lines!(ax, new_time, new_volume)
save(joinpath(output_path, "late_pleistocene_ice_volume.png"), fig)
```
