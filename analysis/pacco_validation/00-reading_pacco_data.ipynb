```python
using NCDatasets, CairoMakie

# Load ice volume
ROOT = pathof("tfm") 
path2data = "my/path/to/data/pacco.nc"
time, volume = NCDataset(path2data)["time"][:], NCDataset(path2data)["Vol"][:]

# Plot ice volume
fig = Figure()
ax = Axis(fig[1, 1])
lines!(ax, time, volume)
save("./ice_volume.png", fig)

# Select time slice
function find_times(t0, tend, time_vector) 
     t0_index = findmin(abs.(t0 .- time_vector))
     tend_index = findmin(abs.(tend .- time_vector))
     return t0_index, tend_index
end

# -- Select only Late Pleistocene
tspan = find_times(-800e3, 0.0, time)
new_time, new_volume = time[tspan[1]:tspan[2]], volume[tspan[1]:tspan[2]]

# Plot
 fig = Figure()
ax = Axis(fig[1, 1])
lines!(ax, new_time, new_volume)
save("./late_pleistocene_ice_volume.png", fig)
```
